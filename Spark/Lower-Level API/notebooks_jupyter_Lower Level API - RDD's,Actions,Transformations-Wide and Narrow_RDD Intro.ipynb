```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName('RDD Intro').master('yarn').getOrCreate()
```

```python
!hadoop fs -ls /tmp/
```

```python
hdfs_path="/tmp/input_spark.txt"
small_rdd=spark.sparkContext.textFile(hdfs_path)
```

```python
small_rdd.collect()
```

```python
small_rdd_clean=small_rdd.map(lambda x:x.split(' '))
small_rdd_clean.collect()
```

```python
small_rdd_clean=small_rdd.map(lambda x:x.split(' ')).map(lambda lst: [each.strip() for each in lst if each !=',']  ).flatMap(lambda x:x)
```

```python
small_rdd_clean.count()
```
