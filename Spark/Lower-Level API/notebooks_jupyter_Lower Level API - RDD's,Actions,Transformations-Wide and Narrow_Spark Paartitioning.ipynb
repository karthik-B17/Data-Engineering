```python
from pyspark.sql import SparkSession
spark=SparkSession.builder\
.appName('Spark_Parallelize_and_Partition')\
.master('yarn')\
.getOrCreate()
```

```python
data = [
"Goku Vegeta Gohan",
"Goku Frieza Goku",
"Vegeta Goku Frieza Gohan",
"Gohan Frieza Goku Goku" ]
```

```python
local_rdd=spark.sparkContext.parallelize(data,2)
```

```python
local_rdd.count()
```

```python
# get num of partition from the rdd as getNumPartition()--This is wrong...?
local_rdd.getNumPartitions()
```
