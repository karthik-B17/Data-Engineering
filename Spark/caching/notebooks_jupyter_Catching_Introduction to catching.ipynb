```python
from pyspark.sql import SparkSession
spark=SparkSession.builder\
.appName('SparkCatching')\
.enableHiveSupport()\
.getOrCreate()
```

```python
customers_rdd=spark.sparkContext.textFile('/tmp/1MB/customers.csv')
```

```python
customers_rdd.take(10)
```

```python
customers_filtered=customers_rdd.filter(lambda row:'Maharashtra' in row )
```

```python
customers_filtered.take(5)
```

```python
customers_mapped=customers_filtered.map(lambda x: (x.split(',')[0],1))
customers_mapped.take(5)
```

```python
customers_reduce=customers_mapped.reduceByKey(lambda x,y:x+y)
```

```python
customers_reduce.take(5)
```

```python
customers_reduce.count()
```
