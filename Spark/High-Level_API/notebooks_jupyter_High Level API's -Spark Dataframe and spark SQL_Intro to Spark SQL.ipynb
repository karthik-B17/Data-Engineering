# DDL Schema


```python
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder\
.appName('use Spark Higher level api-part_2')\
.getOrCreate()
```

```python
ddl_schema="Email STRING,Address STRING,Avatar STRING,Avg_Session_Length DOUBLE,Time_on_App DOUBLE,Time_on_Website DOUBLE,Length_of_Membership DOUBLE,Yearly_Amount_Spent DOUBLE"
```

```python
df=spark.read\
.format('csv')\
.option('header','true')\
.schema(ddl_schema)\
.load('data/Ecommerce_Customers.csv')
```

```python
df.show()
```

```python
df_failfast=spark.read\
.format('csv')\
.option('header','true')\
.schema(ddl_schema)\
.option('mode','FAILFAST')\
.load('data/New_data.csv')
```

```python
df_failfast.show()
```

```python
df_permissive=spark.read\
.format('csv')\
.option('header','true')\
.schema(ddl_schema)\
.option('mode','PERMISSIVE')\
.load('data/New_data.csv')
```

```python
df_permissive.show()
```

```python
df_dm=spark.read\
.format('csv')\
.option('header','true')\
.schema(ddl_schema)\
.option('mode','DROPMALFORMED')\
.load('data/New_data.csv')
```

```python
df_dm.show()
```

```python
spark.stop()
```
