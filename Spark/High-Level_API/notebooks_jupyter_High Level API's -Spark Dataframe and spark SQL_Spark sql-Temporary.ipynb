```python
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder\
.appName('Table Demo')\
.master('Yarn')\
.getOrCreate()
```

```python
! hadoop fs -ls /tmp/1MB*
```

```python
df=spark.read\
.format('csv')\
.option('header','True')\
.option('inferSchema','True')\
.load('/tmp/1MB/customers.csv')
```

```python
df.show(10)
```

```python
spark.sql('show tables').show()
```

```python
df.createOrReplaceTempView('my_first_table')
```

```python
spark.sql('show tables').show()
```

```python
spark.sql("select * from my_first_table limit 5").show()
```

```python
spark_new=spark.newSession()
```

```python
spark_new.sql('show tables').show()
```

```python
# Global Temporary Table
```

```python
df.createOrReplaceGlobalTempView('Global_customers')
```

```python
spark.sql('show tables in global_temp').show()
```

```python
spark.sql('select * from global_temp.global_customers limit 5').show()
```
