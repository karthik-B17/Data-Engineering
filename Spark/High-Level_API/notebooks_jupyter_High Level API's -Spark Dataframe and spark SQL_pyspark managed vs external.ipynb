```python
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder\
.appName('Managed_VS_external')\
.enableHiveSupport()\
.getOrCreate()
```

```python
df=spark.read\
.format('csv')\
.option('header','True')\
.option('inferSchema','True')\
.load('/tmp/1MB/customers.csv')
```

```python
df.createOrReplaceTempView('temp_customers')
```

```python
spark.sql('show tables').show()
```

```python
spark.sql('''
create table managed_customers as select * from temp_customers

''')
```

```python
spark.sql('describe extended managed_customers').show(truncate=False)
```
