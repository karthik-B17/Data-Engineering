```python
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder\
.appName('spark_persistent')\
    .enableHiveSupport()\
.getOrCreate()
```

```python
df=spark.read\
.format('csv')\
.option('header','True')\
.option('inferSchema','True')\
.load('/tmp/1MB/customers.csv')
```

```python
df.show(10)
```

```python
df.write.mode('overwrite').saveAsTable('customer_persistent')
```

```python
spark.sql('show tables').show()
```

```python
spark.sparkContext.applicationId
```

```python
spark.sql('describe extended customer_persistent').show(truncate=False)
```
